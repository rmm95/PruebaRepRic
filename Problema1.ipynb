Una cepa de E. coli tiene una tasa de crecimiento tal que un individuo se duplica cada 30 minutos.
Calcula en cuánto tiempo esta cepa:
a) Llenaría el volumen de la Tierra con su radio (considerandola como una esfera hueca)
b) Cubrir la superficie de la tierra
c) Repite el cálculo para una tasa de crecimiento
 a) Se duplica cada hora
 b) Se duplica cada 1/4 de hora

```python
import math as mt
#El radio de la tierra en km
rT = (6371e3) ** 3
#Raio de E. coli en um
rE = (0.5e-6) ** 3
#Cálculo para saber cuántas caben por despeje de volumen nos queda:
N = rT / rE
#Para el tiempo que tarda en llenar la tierra:
mt.log
t = mt.log2(N)
td = t / 48
print (td, "días")
```
